# RQ3: cross-context comparison
Spearman correlations among frequency, intensity and regularity (median gap in days) in each context, with 95% bootstrap confidence intervals. Units differ between datasets, so only relationships are compared, not raw values.
Note: frequency and the gap between active days are linked by construction (more active days in a fixed period leave less room for long gaps).

```python
import matplotlib as mpl
import matplotlib.pyplot as plt
C=['#2a78d6','#eb6834','#1baf7a','#eda100']
INK='#0b0b0b'; INK2='#52514e'; GRID='#e4e3de'; SURF='#ffffff'
mpl.rcParams.update({'font.family':'DejaVu Sans','font.size':7,'axes.edgecolor':GRID,'axes.labelcolor':INK2,'xtick.color':INK2,'ytick.color':INK2,
 'axes.grid':True,'grid.color':GRID,'grid.linewidth':0.8,'axes.spines.top':False,'axes.spines.right':False,'axes.titlesize':7.5,'axes.titlecolor':INK,
 'axes.titleweight':'bold','axes.titlelocation':'left','lines.linewidth':1.5,'lines.solid_capstyle':'round','legend.frameon':False,'figure.dpi':200,'savefig.dpi':200,
 'savefig.bbox':'tight','figure.facecolor':SURF,'axes.facecolor':SURF,'xtick.major.size':0,'ytick.major.size':0,'legend.fontsize':6.8})

import os, pandas as pd; os.makedirs('figures',exist_ok=True); pd.set_option('display.width',220); pd.set_option('display.max_columns',30)
```

```python
import pandas as pd, numpy as np
o=pd.read_csv('../RQ1/RQ1_OULAD_features.csv')[['frequency','intensity','median_active_day_gap']]
mo=pd.read_csv('../RQ2/RQ2_Moodle_features.csv')[['sessions_per_week','median_events_per_session','median_active_day_gap']]
o.columns=mo.columns=['frequency','intensity','gap']
pairs=[('frequency','gap','Frequency ↔ Regularity (gap)'),('frequency','intensity','Frequency ↔ Intensity'),('intensity','gap','Intensity ↔ Regularity (gap)')]
rng=np.random.default_rng(0); rows=[]
for name,d,B in [('OULAD',o,300),('Moodle',mo,2000)]:
    for a_,b_,lab in pairs:
        rho=d[[a_,b_]].corr('spearman').iloc[0,1]
        bs=[d.iloc[rng.integers(0,len(d),len(d))][[a_,b_]].corr('spearman').iloc[0,1] for _ in range(B)]
        rows.append((name,lab,len(d),rho,*np.percentile(bs,[2.5,97.5])))
R=pd.DataFrame(rows,columns=['context','relationship','n','rho','ci_low','ci_high']); print(R.round(3).to_string(index=False))
```
